```python
from GA import *

yml_file = 'param.yaml'
with open(yml_file) as file:
    param = yaml.load(file, Loader=yaml.FullLoader)

if param is None:
    raise ValueError("Parameter file is empty or not found.")

class arg:
    def __init__(self, **kwargs):
        self.num_of_agents = 10
        self.numOFgenerations = 100
        self.populationSize = 10
        self.agent_idx = None
        self.evolutionTarget = 0
        self.epoch = 0
        self.evolutionType = 'GA'
        self.path = 'TTTTT = test'
        
args = arg()

# crate directory
path = args.path + '/genes/'
if not os.path.exists(path):
    os.makedirs(path)

ea = difEvoInit(vars_and_bounds=param, args=args, geneMin=0)
```

```python
# compare the two parameters
def compare_dicts(d1, d2):
    if isinstance(d1, str) and d1.count(',') > 0:
        d1 = d1.split(',')
    if isinstance(d2, str) and d2.count(',') > 0:
        d2 = d2.split(',')
        
    if isinstance(d1, dict) and isinstance(d2, dict):
        if d1.keys() != d2.keys():
            return False
        for key in d1:
            if not compare_dicts(d1[key], d2[key]):
                print(f'param    org     : {d1[key]}')
                print(f'param reconstract: {d2[key]}')
                print('------')
                return False
        return True
    elif isinstance(d1, list) and isinstance(d2, list):
        if len(d1) != len(d2):
            return False
        for item1, item2 in zip(d1, d2):
            if not compare_dicts(item1, item2):
                return False
        return True
    else:
        return d1 == d2


# get list of genes files
genes = os.listdir(path)
genes = [g for g in genes if g.endswith('.yaml')]

gene_idx = 0

# load a gene  
with open(path + genes[gene_idx]) as file:
    gene = yaml.load(file, Loader=yaml.FullLoader)

# convert gene to parameter
param = {}
_, param = ea.geneTOparam(gene['gene'], param_list=param)

# convert parameter to gene
gene_clone = ea.paramTOgene(param)

# compare the two genes
print('gene:', gene['gene'])
print('gene_clone:', gene_clone)
print('--------------------------------')

param_clone = {}
_, param_clone = ea.geneTOparam(gene_clone, param_list=param_clone)

check_param = compare_dicts(param, param_clone)
if check_param:
    print('param and param_clone are the same')
else:
    print('param and param_clone are different')

print('--------------------------------')
print('param:', param)
print('param_clone:', param_clone)


```
